```python
# ========== Additional Visualizations ==========

# Figure 1: Top 10 Countries by Confirmed Cases - Bar Chart
top10 = data.groupby('Region')['Confirmed'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
top10.plot(kind='bar', color='steelblue')
plt.title('Top 10 Countries by Confirmed COVID-19 Cases (as of Apr 2020)', fontsize=14)
plt.xlabel('Country')
plt.ylabel('Confirmed Cases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('top10_confirmed_cases.png', dpi=150)
plt.show()

# Figure 2: Confirmed vs Recovered - Scatter Chart
top20 = data.groupby('Region')[['Confirmed', 'Recovered']].sum().head(20)
plt.figure(figsize=(10, 6))
plt.scatter(top20['Confirmed'], top20['Recovered'], alpha=0.7, color='orange')
plt.title('Confirmed vs Recovered COVID-19 Cases by Country', fontsize=14)
plt.xlabel('Confirmed Cases')
plt.ylabel('Recovered Cases')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('confirmed_vs_recovered.png', dpi=150)
plt.show()

print("Two charts generated and saved as PNG files!")
```
